### Automatic Keyword Extraction using Sentence Transformers
####Objective
We’ll build a simple semantic keyword extractor using the SentenceTransformer model.
Given any input text, the model will find the most relevant keywords (topics) based on cosine similarity between the text and a predefined list of keywords.

```python
import numpy as np
import torch
from sentence_transformers import SentenceTransformer , util
```

```python
model = SentenceTransformer("all-MiniLM-l6-v2")
```

```python
sent = ["I am good" , "i am fine"]
emb = model.encode(sent)
sim = util.cos_sim(emb[0], emb[1])
sim
```

```python
keywords = [
    "artificial intelligence", "machine learning", "blockchain", "cybersecurity",
    "software engineering", "football", "basketball", "Olympics", "athletics",
    "fitness", "election", "policy", "diplomacy", "legislation", "foreign relations",
    "biology", "physics", "chemistry", "genetics", "astronomy", "healthcare",
    "finance", "economics", "environment", "entertainment", "education", "psychology", "global warming"]
```

### Building the keyword extraction function

#####we will define a function that encodes the input text and comapres it with keyword using cosine similarity. The top few matches (hihest simlarity scores) will be as the most relevant scores.

```python
def extract_top_keywords(content , keywords, top_n =3):
  keyword_emb = model.encode(keywords)
  content_emb = model.encode(content)

  similarities = util.cos_sim(content_emb , keyword_emb).flatten()
  top_indices =torch.topk(similarities , top_n).indices
  top_keywords = [keywords[i] for i in top_indices]
  return top_keywords
```

```python
while True:

    content = input("Input text to get keywords of press(exit): ")

    if content.lower() == "exit":
        break

    else:
        top_keywords = extract_top_keywords(content, keywords, top_n=3)
        print("Top keywords:", top_keywords)
```
